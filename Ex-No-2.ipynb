```python
print("roll no - 24BAD008 ")
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score

```

```python
print("roll no - 24BAD008 ")
df=pd.read_csv("bottle.csv",low_memory=False)
```

```python
print("roll no - 24BAD008 ")
features = ['Depthm', 'Salnty', 'O2ml_L',]
target = 'T_degC'

df = df[features + [target]]
```

```python
print("roll no - 24BAD008 ")
df=df.fillna(df.mean())
```

```python
print("roll no - 24BAD008 ")
X=df[features]
Y=df[target]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
print("roll no - 24BAD008 ")
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size = 0.2,random_state=42)
```

```python
print("roll no - 24BAD008 ")
lr = LinearRegression()
lr.fit(X_train, Y_train)
```

```python
print("roll no - 24BAD008 ")
y_pred = lr.predict(X_test)
print(y_pred)
```

```python
print("roll no - 24BAD008 ")
mse = mean_squared_error(Y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, y_pred)

print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
print("roll no - 24BAD008 ")
plt.figure()
plt.scatter(Y_test, y_pred)
plt.xlabel("Actual Temperature")
plt.ylabel("Predicted Temperature")
plt.title("Actual vs Predicted Water Temperature")
plt.show()
```

```python
print("roll no - 24BAD008 ")
residuals = Y_test - y_pred

plt.figure()
sb.histplot(residuals, kde=True)
plt.xlabel("Residual Error")
plt.title("Residual Error Distribution")
plt.show()
```

```python
print("roll no - 24BAD008 ")
plt.figure(figsize=(8,6))
sb.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
print("roll no - 24BAD008 ")
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, Y_train)
ridge_pred = ridge.predict(X_test)

print("Ridge R²:", r2_score(Y_test, ridge_pred))
```

```python
print("roll no - 24BAD008 ")
df=pd.read_csv("C:\VS code\LICI - 10 minute data.csv")
```

```python
print("roll no - 24BAD008 ")
df["Price_Movement"] = np.where(df["close"] > df["open"], 1, 0)
```

```python
print("roll no - 24BAD008 ")
features = ["open", "high", "low", "volume"]
target = "Price_Movement"

X = df[features]
y = df[target]
X = X.fillna(X.mean())

```

```python
print("roll no - 24BAD008 ")
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
print("roll no - 24BAD008 ")
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

print("roll no - 24BAD008 ")
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
print("roll no - 24BAD008 ")
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("roll no - 24BAD008 ")
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix

print("roll no - 24BAD008 ")
cm = confusion_matrix(y_test, y_pred)

plt.figure()
sb.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

print("roll no - 24BAD008 ")
fpr, tpr, _ = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label="AUC = %.2f" % roc_auc)
plt.plot([0,1], [0,1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
print("roll no - 24BAD008 ")
importance = pd.Series(
    model.coef_[0],
    index=features
)

importance.sort_values().plot(kind="barh")
plt.title("Feature Importance (Logistic Regression)")
plt.show()
```

```python
print("roll no - 24BAD008 ")
from sklearn.model_selection import GridSearchCV

param_grid = {
    "C": [0.01, 0.1, 1, 10],
    "penalty": ["l1", "l2"]
}

grid = GridSearchCV(
    LogisticRegression(
        solver="liblinear",
        max_iter=5000
    ),
    param_grid,
    cv=5
)

grid.fit(X_train, y_train)
```

```python
print("roll no - 24BAD008 ")
y_pred_best = grid.best_estimator_.predict(X_test)

print("Optimized Accuracy:",
      accuracy_score(y_test, y_pred_best))
```
